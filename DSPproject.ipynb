```python
def perform_sampling(lewandowski_frequency, pedri_duration, gavi_sampling_rate_continuous, terstegen_sampling_rate_sampled):

    # Continuous wave
    frenkie_time_continuous = np.linspace(0, pedri_duration, int(pedri_duration * gavi_sampling_rate_continuous), endpoint=False)
    raphinha_continuous_signal = np.sin(2 * np.pi * lewandowski_frequency * frenkie_time_continuous)

    # Sampled signal
    ansu_time_sampled = np.linspace(0, pedri_duration, int(pedri_duration * terstegen_sampling_rate_sampled), endpoint=False)
    dembele_sampled_signal = np.sin(2 * np.pi * lewandowski_frequency * ansu_time_sampled)

    # Plot continuous and sampled signals
    plt.figure(figsize=(10, 6))
    plt.plot(frenkie_time_continuous, raphinha_continuous_signal, label="Continuous Signal", color='blue', alpha=0.7)
    plt.stem(ansu_time_sampled, dembele_sampled_signal, linefmt='r-', markerfmt='ro', basefmt=' ', label="Sampled Signal")
    plt.title("Sampling a Sine Wave")
    plt.xlabel("Time (seconds)")
    plt.ylabel("Amplitude")
    plt.legend()
    plt.grid()
    plt.show()

    return frenkie_time_continuous, raphinha_continuous_signal, ansu_time_sampled, dembele_sampled_signal


# Quantization
def quantize_signal(signal, num_levels):

    xavi_max_val = np.max(signal)
    iniesta_min_val = np.min(signal)
    puyol_step_size = (xavi_max_val - iniesta_min_val) / (num_levels - 1)  # Step size for quantization
    deco_quantized_signal = np.round((signal - iniesta_min_val) / puyol_step_size) * puyol_step_size + iniesta_min_val
    return deco_quantized_signal


#Plot 2l Quantized Signal

def plot_quantized_signal(ansu_time_sampled, dembele_sampled_signal, deco_quantized_signal):

    plt.figure(figsize=(10, 6))

    # Plot sampled signal
    plt.stem(ansu_time_sampled, dembele_sampled_signal, linefmt='r-', markerfmt='ro', basefmt=' ', label="Sampled Signal")

    # Plot quantized signal
    plt.step(ansu_time_sampled, deco_quantized_signal, where='mid', color='green', label="Quantized Signal", linewidth=2)

    plt.title("Quantization of Sampled Signal")
    plt.xlabel("Time (seconds)")
    plt.ylabel("Amplitude")
    plt.legend()
    plt.grid()
    plt.show()

```

```python
# ===================== Main Code ===================== #

# parameters 2l Sampling
lewandowski_frequency = 5
pedri_duration = 1
gavi_sampling_rate_continuous = 1000  # High sampling rate for "continuous" signal 34an 2l aliasing (distortion)
terstegen_sampling_rate_sampled = 10  # 34an 24oof 2l signal 10 points kol sec


frenkie_time_continuous, raphinha_continuous_signal, ansu_time_sampled, dembele_sampled_signal = perform_sampling(
    lewandowski_frequency, pedri_duration, gavi_sampling_rate_continuous, terstegen_sampling_rate_sampled
)

# parameters 2l Quantization
num_quantization_levels = 4 # 4 levels


deco_quantized_signal = quantize_signal(dembele_sampled_signal, num_quantization_levels)

plot_quantized_signal(ansu_time_sampled, dembele_sampled_signal, deco_quantized_signal)
```

```python
def calculate_quantization_error(original_signal, quantized_signal):


    quantization_error = original_signal - quantized_signal
    return quantization_error
```

```python
# ========== Functions for Noise Reduction ========== #


from scipy.signal import butter, filtfilt
from scipy.signal import medfilt

def apply_low_pass_filter(signal, sampling_rate, cutoff_frequency):
    nyquist_rate = sampling_rate / 2  # nyquist rate 50/2 = 25
    normalized_cutoff = cutoff_frequency / nyquist_rate  # normalized cutoff frequency
    if not (0 < normalized_cutoff < 1):
        raise ValueError("Cutoff frequency must be between 0 and the Nyquist frequency.")
    # design 2l filter
    b, a = butter(N=4, Wn=normalized_cutoff, btype='low')
    filtered_signal = filtfilt(b, a, signal)  # fit 2l filter
    return filtered_signal

def apply_median_filter(signal, kernel_size):



    filtered_signal = medfilt(signal, kernel_size=kernel_size)
    return filtered_signal
```

```python
def plot_filtered_signals(original_signal, smoothed_signal, low_pass_filtered_signal, median_filtered_signal, time_sampled):
    """
    Plot the original, smoothed, low-pass filtered, and median-filtered signals for comparison.
    """
    plt.figure(figsize=(12, 8))

    # Original signal
    plt.plot(time_sampled, original_signal, label="Original Quantized Signal", color='blue', alpha=0.7)

    # Smoothed signal
    plt.plot(time_sampled[:len(smoothed_signal)], smoothed_signal, label="Smoothed Signal (Moving Avg)", color='green', linewidth=2)

    # Low-pass filtered signal
    plt.plot(time_sampled[:len(low_pass_filtered_signal)], low_pass_filtered_signal, label="Low-Pass Filtered Signal", color='orange', linestyle='--', linewidth=2)

    # Median filtered signal
    plt.plot(time_sampled[:len(median_filtered_signal)], median_filtered_signal, label="Median Filtered Signal", color='red', linestyle='-.', linewidth=2)

    plt.title("Signal Comparison: Original, Smoothed, Low-Pass Filtered, and Median-Filtered")
    plt.xlabel("Time (seconds)")
    plt.ylabel("Amplitude")
    plt.legend()
    plt.grid()
    plt.show()
```

```python
#High pass filter

def apply_high_pass_filter(signal, sampling_rate, cutoff_frequency):

    # Fourier Transform
    frequency_spectrum = np.fft.fft(signal)
    frequencies = np.fft.fftfreq(len(signal), d=1/sampling_rate)


    high_pass_filter = np.abs(frequencies) > cutoff_frequency #remove low freqs

    #apply filter 3la 2l fourier spectrum
    filtered_spectrum = frequency_spectrum * high_pass_filter

    # تحويل الإشارة مرة أخرى إلى نطاق الزمن
    filtered_signal = np.fft.ifft(filtered_spectrum).real

    return filtered_signal
```

```python
def plot_high_pass_filtered_signal(original_signal, filtered_signal, time_points):

    # Ensure time_points matches filtered_signal length
    time_points = time_points[:len(filtered_signal)]

    plt.figure(figsize=(12, 8))
    plt.plot(time_points, original_signal[:len(filtered_signal)], label="Original Smoothed Signal", color='blue', alpha=0.7)   #original smoothed signal
    plt.plot(time_points, filtered_signal, label="High-Pass Filtered Signal", color='red', linestyle='--')     #high pass filter applied
    plt.title("High-Pass Filtered Signal")
    plt.xlabel("Time in seconds")
    plt.ylabel("Amplitude")
    plt.legend()
    plt.grid()
    plt.show()
```

```python
def apply_band_pass_filter(signal, sampling_rate, low_cutoff, high_cutoff, filter_order=4):
    """
    Apply a band-pass filter to a signal.

    Parameters:
         Input signal to filter.
        Sampling rate of the signal
        Lower cutoff frequency for the band-pass filter
        Upper cutoff frequency for the band-pass filter

    """
    nyquist = 0.5 * sampling_rate
    low = low_cutoff / nyquist
    high = high_cutoff / nyquist
    b, a = butter(filter_order, [low, high], btype='band')
    return filtfilt(b, a, signal) # b return as an array of points yastaa
```

```python
def apply_band_reject_filter(signal, sampling_rate, low_cutoff, high_cutoff, filter_order=4):

    nyquist_rate = sampling_rate / 2.0
    low = low_cutoff / nyquist_rate
    high = high_cutoff / nyquist_rate
    b, a = butter(filter_order, [low, high], btype='bandstop')  # Bandstop filter
    filtered_signal = filtfilt(b, a, signal)
    return filtered_signal
```

```python
def plot_band_reject_filtered_signal(original_signal, filtered_signal, time, low_cutoff, high_cutoff):

    min_length = min(len(original_signal), len(filtered_signal), len(time))
    time = time[:min_length]
    original_signal = original_signal[:min_length]
    filtered_signal = filtered_signal[:min_length]

    plt.figure(figsize=(10, 6))
    plt.plot(time, original_signal, label="Original Signal", color="blue")
    plt.plot(time, filtered_signal, label=f"Band-Reject Filtered Signal ({low_cutoff}-{high_cutoff} Hz)", color="brown", linestyle="--")
    plt.title(f"Band-Reject Filtered Signal ({low_cutoff}-{high_cutoff} Hz)")
    plt.xlabel("Time in (s)")
    plt.ylabel("Amplitude")
    plt.legend()
    plt.grid(True)
    plt.show()

```
